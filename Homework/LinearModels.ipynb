## Quiz Questions

```python
# Seed the pseudo-random number generator
np.random.seed(12)

# Sample size
n=100

# Create data
wcc = np.round(np.random.normal(15, 5, n), 0)
crp = (wcc * 2) + np.round(np.random.normal(0, 10, n), 0)

lipase = wcc + crp + np.round(np.random.normal(2, 10, n), 0)

# Pandas dataframe obejct
df = DataFrame(
    {'WCC':wcc,
     'CRP':crp,
     'Lipase':lipase}
)
```

1. ##### *Create multiple linear regression model with `WCC` and `CRP` as predictors of `Lipase`.*


```python
# Create multiple linear regression model with `WCC` and `CRP` as predictors of `Lipase`
y, X = dmatrices('Lipase ~ WCC + CRP', data=df, return_type='dataframe')
model = sm.OLS(y, X)
results = model.fit()
print(results.summary())
```

2. ##### *Comment on the $F$ statistic.*

The $F$-statistic is large, meaning the results can vary by up to 157.3 points.

3. ##### *Comment on the individual coefficients and their $p$ values.*

The coefficient for `WCC`is positive but small, meaning the correlation is weak. THe coefficient for `CRP` is also positive and above `1`, so the correlation is very strong. The P values for both `WCC` and `CRP` are below $0.05$, so the recorded data is likely not due to chance. The $p$-value for `CRP` beinge extremely statistically significant.

4. Comment on the $R^2$ value.

The $R$-squared value is adjusted to 0.759 to take into account the multiple dependent variables used in the linear regression model. The adjusted $R$-squared value shows that $75.9$% of the variance of the dependent variable is explained by the independent variables in the model.
