##### 2. *Determine the data type of the output of the code `1+2.` and `1+2`. Explain why the data types are different.*

```python
type(1+2.)
```

```python
type(1+2)
```

 The code `1+2.` is a float data type because it includes a decimal point, but the code `1+2` is an integer because it omits the decimal point. This difference is that the float data type approximates the value beyond the decimal point while integers are strictly whole numbers.

##### 4. *Create the variable `i` by assigning the value $1$ to it. Create a loop that prints the value of $i^2$ to the screen. Increment the value of `i` by $1$ until the value of $i^2$ exceeds $1000$. Print the value assigned to `i` after the loop has completed. Explain why the solution is `i=32`, where $32^2$ is larger than $1000$.*

```python
i = 1

while i**2 < 1000:
    i += 1
print(i)
```

`i=32` is the solution because when `i` reaches $32$, the result of $i^2$ is $1042$, which is when it finally exceeds $1000$ and prints $32$.
