### Import Packages

```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
%config InlineBackend.figure_format = 'retina'
```

```python
sns.set_style(style='whitegrid')
```

```python
df = pd.read_csv('https://github.com/juanklopper/TutorialData/blob/main/heart_failure.csv?raw=true')
```

## Task 1

<mark>*Task:*</mark> Create a scatter plot to visualize the correlation between the `platelets` and `serum_sodium` variables, colored by the unique elements (classes or levels) of the `diabetes` variable. Add an appropriate title and axes labels.

```python
sns.scatterplot(
    data=df,
    x='platelets',
    y='serum_sodium',
    hue='diabetes'
).set(
    title='Platelets and serum sodium',
    xlabel='Platelets [platelets/mL]',
    ylabel='Serum sodium [mEq/L]')
```

## Task 2

<mark>*Task:*</mark> Generate a histogram to visualize the distribution of the `platelets` variable. Calculate the minimum and the maximum value for the variable and use the `range` function to create a intervals of $50000$. Note that the ticks value on the horizontal axis seems overcrowded. Try to use the argument `y` instead of `x`. This will create a horizontal histogram, with more space for the tick values.

```python
df.platelets.min()

df.platelets.max()

sns.displot(
    df,
    y='platelets',
    bins=range(0, 850000,50000)
    ).set(
    title='Platelet distribution',
    ylabel='Platelets [platelets/mL]',
    xlabel='Frequency')
```

## Task 3

<mark>*Task:*</mark> Another categorical variable can be added using the `hue` argument and setting a categorical variable as argument. Create a bar plot using the `diabetes` variable as in the figure above, but add the `death` variable as well.

1. ##### *How do you import the Seaborn package in Python?*

2. ##### *How do you create a scatter plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*

3. ##### *How do you create a line plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*
    You can create the line plot with `seaborn.lineplot(data=df, x=x, y=y)`.
---
4. ##### *How do you create a histogram in Seaborn for a variable `x` in a DataFrame `df`?*
    You can create a histogram with `seaborn.histplot(df.x)`.
---
5. ##### *How do you create a stacked histogram in Seaborn for variables `x` and 'hue' `y` in a DataFrame `df`?*
    You can create a stacked histogram with `seaborn.histplot(data=df, x='x', hue='y', multiple='stack')`.
---
6. ##### *How do you create a heatmap in Seaborn for a variable `x` and `y` in a DataFrame `df`?*
    You can create a heatman with `sns.heatmap(df[['x', 'y']].corr())`.

7. ##### *How do you create a box plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*
    You can create a box plot with `sns.boxplot(data=df, x='x',  y='y')`.
---
8. ##### *How do you create a violin plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*

9. ##### *How do you create a bar plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*

10. ##### How do you add a title to a Seaborn plot assigned to the variable `ax` with the title 'My Title'?

11. ##### *How do you save a Seaborn plot `ax` to a file named 'output.png'?*

12. ##### *How do you set the `x` and `y` axis labels to 'X Label' and 'Y Label' respectively for a Seaborn plot ax?*

15. ##### *How do you add a regression line to a scatter plot in Seaborn for variables `x` and `y` in a DataFrame `df`?*
