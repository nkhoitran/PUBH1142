## Task 1

<mark>*Task:*</mark> Determine the sample space elements, the frequency, and the relevant frequencies of the `diabetes` variable.

```python
df = pd.read_csv("https://raw.githubusercontent.com/juanklopper/TutorialData/main/heart_failure.csv")
df.diabetes.unique()
```

```python
df.diabetes.value_counts()
```

```python
df.diabetes.value_counts(normalize=True)
```

## Task 2

<mark>*Task:*</mark> Generate a contingency table, indicating the joint percentages of the `anaemia` variable for each of the classes in the `death` column.

```python
pd.crosstab(df.anaemia, df.death, normalize=True) * 100
```

## Task 3

<mark>*Task:*</mark> Determine the median `ejection_fraction` for each of the classes in the `anaemia` variable.

```python
df.groupby('anaemia')['ejection_fraction'].median()
```

## Task 4

<mark>*Task:*</mark> Determine the variance in `ejection_fraction` for each of the classes in the `anaemia` variable.

```python
df.groupby("anaemia").ejection_fraction.var(ddof=1)
```
