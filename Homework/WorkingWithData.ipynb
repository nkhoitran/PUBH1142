## Task 4

<mark>*Task:*</mark> Create a variable named selection that will select for all the observations who never smoked, as a heart rate, `HR`, less than $80$, and a systolic blood pressure, `sBP`, of less than $120$. Use this selection to create a numpy $n$-dimensional array of the `Age` and triacylglycerol, `TAG`, variable values.

```python
selection = (df.Smoke==0) & (df.HR < 80) & (df.sBP < 120)
df.loc[selection, ['Age', 'TAG']].to_numpy
```

## Task 5

<mark>*Task:*</mark> Create a new variable named `AgeHR` that is the product of the values in the variables `Age` and `HR`.

```python
df['AgeHR'] = df.Age * df.HR
```

## Task 6

<mark>*Task:*</mark> Create a new variable in the dataframe called `HRLevel` from the `HR` column with intervals ($0,65$], ($65,78$], ($78,1000$) with labels `low`, `normal`, and `high` respectively.

```python
df['HRLevel'] = pd.cut(df.HR, bins = [0, 65, 78, 1000], right=True, labels=['low', 'normal', 'high'])
df.HRLevel[:5]
```
