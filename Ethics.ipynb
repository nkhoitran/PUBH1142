## Misuse of Health Data

 *<mark>Task 1:</mark> Find another example of the misuse of health data. Give details about the event and discuss the ethical issues involved. What could have been done to prevent this from happening?*

At Tuskegee University, researchers collected data from local residents to investigate syphilis from 1932 to 1972. However, as they monitored and tracked the development of “latent syphilis” in 399 black men, they denied them proper treatment with placebos in order to observe the extent of the disease’s symptoms, leaving patients “blind or insane or experienc[ing] other severe health problems due to their untreated syphilis.”[^1] Moreover, there was no evidence that researchers obtained consent, breaching privacy and rights. Participants were informed that the study was a “government health care program,” but there was no intention in properly treating them, even after penicillin became an available and widespread treatment for syphilis.[^2] As the study continued, they ultimately facilitated the spread of syphilis by allowing the untreated syphilis to spread to other women through their subjects. As a result of the malpractice, 128 people had died due to absent treatments, and syphilis had infected 59 people.[^3] The Tuskegee syphilis study is an infamous example of malpractice in health data, and it marked the necessity for ethical bounds for experiments, proper consent from participants, and adequate public information about the study that could have prevented such atrocities from occurring.
The study at Tuskegee was primarily unethical because they kept vital treatment away from vulnerable patients, taking advantage of their disease to further the researcher’s study on syphilis at the expense of their subjects’ lives. Designated supervisors for studies could work to prevent patient exploitation, as they can assess researchers and intervene when actions become malpractice or harmful. Alongside an association, universal rules for ethical bounds can help gauge the ethical standpoint of certain experiments, as a review board was created for the Tuskegee study, but ruled for the study to continue.[^4] The syphilis study also highlighted how researchers overstepped people’s consent, which could have been prevented with proper laws and barriers that protect people’s rights to choice. Requirements for a study to possess evidential consent would have caused researchers to be conscious and meticulous about considering people’s rights, preventing incidents that manipulate communities. Lastly, proper community knowledge and publication of the study’s dangerous extents would have prevented the research from surviving for 40 years, as public pressure and criticism would have halted the study before it brought harm to any participants. The study ended in 1972 because Peter Buxton, a disease investigator, had informed reporters of the malpractice, causing the research to receive widespread and loud public backlash that forced the study to conclude.[^5] Efforts to publicize the full scope of the study would impede longlasting research in the community, as outcry and rebellion would have been present earlier in the study.
Although the Tuskegee syphilis study was a tragedy that caused public mistrust in medicine and health services, it spread awareness for the importance of maintaining proper ethics in research and treatment. Affected participants received a combined court settlement of $10,000,000 dollars, and public health fields created efforts to address the history of the study to understand the causes, shortcomings, and results to better equip future research to cultivate health equality.[^6] Prior installment of ethical guidelines for studies, required proof of consent, and sufficient public information would have kept the Tuskegee community safe from abuse, demonstrating the need to address and consider the risks involved in certain studies, as well as measures to prevent unethical approaches for personal gain.

## Breaches in Health Data Security

 *<mark>Task 2:</mark> Find another example of bias in health data. Describe the event and the consequences. What could have been done to prevent this from happening?*

A Thai private hospital in 2024 failed to properly monitor and dispose of personal medical documents, which led to the unethical storage of over 1,000 medical records by a small disposal company and the papers being recycled and circulated within the community. In the Ubon Ratchathani province, the hospital “outsourced the file destruction work to a small, family-run business without monitoring the process,”[^1] causing the catastrophic leak to go unnoticed until the papers began appearing publicly, notably as food wrappers from small businesses with patient notes, medical diagnoses, and “‘sensitive personal data’” under Section 26 of the Personal Data Protection Act.”[^2] The incident led to the Personal Data Protection Committee in Thailand fining the private hospital the equivalent of 37,000 U.S. Dollars for violating Thailand’s Personal Data Protection Act and set the precedent for the initiative for proper action with procedures regarding sensitive personal data.[^3] Although small compared to other health data breaches, the Thai hospital’s miscare of private health documents highlighted the need for careful surveillance and proper handling of health data that could have prevented the spread of private information from over 1000 individuals.
As the hospital outsourced its medical records, it failed to document and record the data’s destruction, which allowed the data to be easily misplaced and mishandled in a manner that compromised patients’ security and privacy. If the hospital had established proper maintenance that observed the record’s destruction, the unrightful storage of the documents by the disposal company would have been recognized, enabling the hospital to act before the breach occurred. Moreover, the hospital could have also taken measures to obscure and anonymize the data before outsourcing the personal data to the third party business, adding layers of protection to the disposal process. Although not completely preventing the circulation of the medical records, redacting identifiable factors in the medical records beforehand can minimize the risk and danger of such incidents, protecting personal privacy and preventing identification of individuals instead.
Managing personal health data requires meticulous organization and logging in every step of handling medical documents, as allowing such records to become public jeopardizes people’s privacy and safety. By implementing precautions and detailed practices, hospitals can lower the risk of private data reaching the public’s access and keep patients anonymous and secure. Although this case applies to physical, paper records, these same methods can also be applied to other applications of health data, such as digital data. By recording where all copies of online records are stored and anonymizing defining information, companies can minimize the risk of online libraries and files from breaches and organized attacks. Through reflecting on these incidents, hospitals can apply the necessary precautions and safety measures for communities to maintain their trust in medical facilities.

## Bias in Health Data

 *<mark>Task 3:</mark> Find another example of bias in health data. Describe the event and the consequences. What could have been done to prevent this from happening?*

In a 2019 study, researchers demonstrated how a software, utilized by insurance companies and hospitals aimed to identify high-risk individuals, admitted healthier white patients to “high-risk management programs”[^1] ahead of less healthy black patients. The bias came from the algorithm correlating higher-cost patients with medical needs, and because white individuals tended to be more costly, despite being healthier, than black people, the bias caused hospitals to unintentionally exclude higher priority black people with chronic or complex diseases from programs that would help maintain routine medication.[^2] However, by factoring in the health disparities that skewed the algorithm and introducing more aspects to influence judgement, researchers were able to remedy the bias for the software to work effectively and as intended.[^3] Due to underlying systemic differences in healthcare, the software showed inherent bias towards white people when determining who automatically qualified for health programs, undermining health equality and ignoring individuals desperately needing medical aid. Shortcomings and flaws in software, as well as applications from skewed results, underscore the need for health disparity awareness and constant consideration of inequality in data to protect marginalized groups from sustained inequality and promote health equity.
Although such circumstances can result from accidental oversights in data, intentional efforts for researchers and hospitals to be actively aware of ingrained bias facilitates an environment that is constantly identifying and eliminating prejudice from health data. Institutions can implement checks for bias similar to the study that identified the software’s racial bias beforehand in order to prevent situations where research or algorithms result in marginalized groups in health care. Education and informative interventions about health disparities would keep researchers and developers conscious of the possible influences on data, allowing people to factor distortions before the implementation of software and research. Moreover, one can prevent biased health data by incorporating multiple factors to counteract the effects of skewed data. In the case of the mentioned algorithm, by incorporating “other variables to predict patient risk, such as costs that could be avoided by preventative care,” the researchers removed enough bias from the software, allowing black admissions to high risk programs to increase from 18% to 47%.[^4] If institutions can promote practices that combat against bias, such as considering multiple factors, early on, then hospitals can prevent scenarios that leave certain groups vulnerable to circumstance and discrimination.
Analyzing data allows hospitals to address needed areas in health and combat prevalent harms threatening people’s wellbeing. However, ignoring cemented bias in the health care system and the pervasiveness of health disparities enables existing inequality to persist. By both recognizing and understanding the presence of possible injustice and unfairness in these systems, communities can work to prevent their influence in public health.

## Anonymization of Health Data

 *<mark>Task 4:</mark> Find an example of anonymization of health data. Describe the technique used.*

Researchers utilize anonymization to protect the privacy and security of individuals by obscuring identifiers in data; however, the necessity to anonymize health data introduces new roadblocks when addressing patient information. One such issue is linking the same patient in data between institutions and hospitals, as some anonymization techniques do not have means to identify the same individual without exposing the person’s identity. Providing a means to do so can be useful when patients transfer to different hospitals or are under different physicians at different times of their life.[^1] Therefore, hospitals employ hashing and salting as a technique to assign a unique alphanumerical code to a single patient based on sending their identifiable data through a one-way encryption algorithm alongside a chosen string of characters to complicate the process, guaranteeing the data from being reversed engineered. Afterwards, identifiable data is redacted and are replaced by the new, separated marker for all documents pertaining to the patient, allowing studies to associate documents from different sources that apply to the same patient, if the same methods or algorithm was used to mark them, without acknowledging nor identifying the real person from such data. This method has been used to anonymize health data in a conclusive online study that provided utility and functionality, supplementing flexibility and convenience to physicians and hospitals:
>By using a hash-and-salt algorithm, we were able to conduct a multisite study through which we retrospectively gathered 10 years’ worth of data about patients across multiple physicians and family medicine clinics while keeping patient data linked, despite having removed personally identifiable information.[^2]

Through feeding algorithms a patient’s name, birthdate, medical card number, and other key identifiers, special codes, which take “4.54×1017 combinations” of reverse engineering, and even more with salting “chosen or generated [characters] that adds trillions more years”[^3] to breach attempts, can both protect people from identity fraud and facilitate research.
