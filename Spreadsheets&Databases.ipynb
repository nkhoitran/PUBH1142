## Quiz Questions

1. ##### *What is the difference between a spreadsheet and a database?*
    A spreadsheet is a program that organizes data in a columns and rows comprised of cells that can hold individual datum to perform calculations, create charts, and provide simple analyses. A database however collects larger amounts of data systematized for complex access and comparison. Databases take more effort to organize and prepare, but allows for more complicated and versatile analyses with more efficiency and tools.
---
2. ##### *What are the advantages and disadvantages of using a spreadsheet for health data management?*
    Spreadsheets are quick and simple tools that are effective with small datasets, allowing for a quick, convenient analysis, and are easy to pickup and learn. However, spreadsheets become inefficient with larger datasets and complex analyses, having limited data capacity, capabilities and integrity.
---
3. ##### *What are the advantages and disadvantages of using a database for health data management?*
    Databases can handle large and complicated datasets while maintaining integrity and supplying a variety of tools for analysis. They provide different ways to compare different sets and variables, allowing for stronger and more useful analysis integral for healthcare. Nevertheless, databases take time and planning to setup, becoming inconvenient when facing urgency or smaller, less complex datasets, and require knowledge in database design and SQL to properly utilize.
---
4. ##### *What are some examples of tasks that can be performed using a spreadsheet?*
    Spreadsheets can apply conditions and rules to formatting for a cleaner, user-friendly interface that is easy to visually analyze. They can perform basic calculations and filter for certain criteria for basic analyses, and spreadsheets can help visualize that data by creating charts and tables. Moreover, spreadsheets can be shared through CSV files, as all spreadsheet programs can import and export data in that format.
---
5. ##### *What are some examples of tasks that can be performed using a database?*
    Databases can reduce redundancy in data through normalization to boost data integrity through consistency by eliminating anomalies. Furthermore, because it can hold larger datasets, it can utilize its comparison tools to identify relationships between multitudes of tables and different health information, becoming pivotal to research and efficient analysis. Databases can also allow for individual user access with simultaneous data manipulation at once, allowing for easier accessibility between multiple people, and it can manage user-level access alongside more advance security to bolster the information's safety and protection.
---
6. ##### *What are some examples of tasks that can be performed using both a spreadsheet and a database?*
    Both a spreadsheet and a database can organize datasets into tables and perform data manipulations to analyze simple aspects such as averages, sums, and correlations. They can both provide analysis tools such as pivot tables and charts to visualize correlations between two variables.
---
7. ##### *What are some examples of tasks that can be performed using neither a spreadsheet nor a database?*
    Neither a spreadsheet nor database can provide a solutions to issues with a given result, as they aim only to organize, analyze, and visualize datasets. They are integral to research, patient care, and policies, but they simply provide people with information to make evidence-based decisions, as individuals still must interpret and apply results and reports to the tackled scenario.
---
8. ##### *What are some examples of tasks that can be performed using a spreadsheet but not a database?*
    Some examples of tasks a spreadsheet can perform but not a database is quick and easy setup for small datasets and analyses. Spreadsheets can be quickly created and learned for on-demand comparison and visualization, which databases struggle to compete with as they take more time and planning to ready.
---
9. ##### *What are some examples of tasks that can be performed using a database but not a spreadsheet?*
    Databases can store larger amounts of data and employ higher levels of data relation, integrity, accessibility, security, and capabilities than spreadsheets. They are able to eliminate anomalies in data for more credible analyses through normalization, and they can connect relations between different assortments of variables and datasets for more applicable analyses.
---
10. ##### *What are some examples of tasks that can be performed using both a spreadsheet and a database?*
    Together, spreadsheets and databases can analyze datasets depending on the context: providing quick, easy spreadsheets with visualization and analyses for smaller sets, or long-term, versatile databases for long-lasting, useful analysis of larger datasets. They provide researchers and policy-makers with information essential for improving patient care and store data in secure, organized manners.
